```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
p=0.3
data = np.random.binomial(1,p,1000)
sns.histplot(data, stat="density", discrete=True)
plt.title("Bernoulli Distribution (p={p})")
plt.xlabel("Outcomes (0=failure,1=success)")
plt.ylabel("Density")
plt.xticks([0.10])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
n=10
p=0.3
data = np.random.binomial(n,p,1000)
sns.histplot(data,kde=True, stat="density", bins=n+1)
plt.title("Bernoulli Distribution (n={n},p={p})")
plt.xlabel("Number of Success")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
iam = 4
data = np.random.poisson(iam,100)
sns.histplot(data, kde = False, stat = "density", bins = range(min(data), max(data)+1))
plt.title(f"Possion Distribution (lambda={iam})")
plt.xlabel("Number of events")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
mu= 0
sigma = 1
data = np.random.normal(mu,sigma,1000)
sns.histplot(data, kde = True, stat = "density")
plt.title(f"Normal Distribution (mean={mu}, std ={sigma})")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
low = 0
high = 1
data = np.random.uniform(low,high,1000)
sns.histplot(data, kde = True, stat = "density")
plt.title(f"Uniform Distribution ({low},{high})")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
iam = 1.5
data = np.random.exponential(1/iam,1000)
sns.histplot(data, kde = True, stat = "density")
plt.title(f"Exponential Distribution (lambda=1.5)")
plt.xlabel("Time")
plt.ylabel("Density")
plt.show()
```
